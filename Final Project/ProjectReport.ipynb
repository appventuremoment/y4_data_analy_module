### Checking Values (Formatting and Sensibility) <a id='checkvalue'/>

First, we repalce any null values since they are most likely from previous years when the question was not asked. We then find the largest values.

```python
finaldf.reason_for_learning.replace(np.nan, 'Not answered', inplace = True)
finaldf.reason_for_learning.value_counts().nlargest(10)
```

From there, we can group every other reason that does not have a significant frequency into the others category.

```python
finaldf.loc[~finaldf.reason_for_learning.isin(list(finaldf.reason_for_learning.value_counts().nlargest(8).index)), 'reason_for_learning'] = 'Others'
finaldf.reason_for_learning.value_counts()
```

For the next four columns, they are all of similar nature. Since some of them are strings (the names) and some of them are integers (number of resources the respondent found helpful), we will split each of the strings by ',' and find the len of the split array. This will give us the number of each resource the respondent has found helpful, which is the format we want.

```python
#course_learn
finaldf.loc[~finaldf.applymap(np.isreal).course_learn, 'course_learn'] = finaldf[~finaldf.applymap(np.isreal).course_learn].course_learn.str.split(',').str.len()
print(list(finaldf.course_learn.unique()))
#event_in_person_learn
finaldf.loc[~finaldf.applymap(np.isreal).event_in_person_learn, 'event_in_person_learn'] = finaldf[~finaldf.applymap(np.isreal).event_in_person_learn].event_in_person_learn.str.split(',').str.len()
print(list(finaldf.event_in_person_learn.unique()))
#podcast_learn
finaldf.loc[~finaldf.applymap(np.isreal).podcast_learn, 'podcast_learn'] = finaldf[~finaldf.applymap(np.isreal).podcast_learn].podcast_learn.str.split(',').str.len()
print(list(finaldf.podcast_learn.unique()))
#youtube_learn
finaldf.loc[~finaldf.applymap(np.isreal).youtube_learn, 'youtube_learn'] = finaldf[~finaldf.applymap(np.isreal).youtube_learn].youtube_learn.str.split(',').str.len()
print(list(finaldf.youtube_learn.unique()))
```

For the hours_per_week column, since we do not want to deal with floats, we need to first convert all nan values to any integer/float so that we can convert the whole column to integers, thereby removing any decimal places, before converting the interger/float value chosen back to nans.

```python
finaldf.hours_per_week = (pd.to_numeric(finaldf.hours_per_week, errors = 'coerce'))
finaldf.hours_per_week.replace(np.nan, 180, inplace = True)
finaldf.hours_per_week = finaldf.hours_per_week.astype(int)
finaldf.hours_per_week.replace(180, np.nan, inplace = True)
```

Next, assuming a week has 7 days with 24 hours each, and each person has a generous estiamte of no more than a maximum of half a day each day to spend on learning how to code, that gives the maximum possible value of 84 hours a respondent can spend on learning how to code each week, which we will round down to 80 since this is a very generous estimate. We will also round up all values between 0 and 1 to 1 hour, since working in whole number is more convenient.

```python
finaldf.loc[(finaldf.hours_per_week > 0) & (finaldf.hours_per_week < 1), 'hours_per_week'] = 1
finaldf.loc[(finaldf.hours_per_week > 80), 'hours_per_week'] = finaldf.hours_per_week.median()
finaldf.hours_per_week.unique()
```

From the value_count of the months_experience column and from common sense, we can deduce that answers outside the range of 0-0.05 or above 600 months do not make sense. The number of such occurrences are low, so they can be replaced with the median.

```python
finaldf.months_experience = (pd.to_numeric(finaldf.months_experience, errors = 'coerce'))
finaldf.loc[(finaldf.months_experience > 0) & (finaldf.months_experience < 0.05) | (finaldf.months_experience > 600), 'months_experience'] = finaldf.months_experience.median()
```

From the value_count of the money_spent_learning, we can deduce that answers in the range of 1 to 30000 dollars makes the most sense. The number of such occurrences outside that range is low, so they can be replaced with the median.

```python
finaldf.money_spent_learning = (pd.to_numeric(finaldf.money_spent_learning, errors = 'coerce'))
finaldf.loc[(finaldf.money_spent_learning > 0) & (finaldf.money_spent_learning < 1) | (finaldf.money_spent_learning > 30000), 'money_spent_learning']  = finaldf.money_spent_learning.median()
```

Looking through nlargest, we find that for 'working in nonprofit', it is repeated with different formatting and that 'Freelancing' is repeated with trailing whitespaces. We fix those formatting issues and categorise the rest into a the unsure/others category. 

```python
finaldf.pref_company_type.replace(np.nan, 'Not answered', inplace = True)
finaldf.loc[finaldf.pref_company_type.str.contains('Work for a nonprofit'), 'pref_company_type'] = 'Work for a non profit'
finaldf.pref_company_type = finaldf.pref_company_type.str.strip()
finaldf.loc[~finaldf.pref_company_type.isin(list(finaldf.pref_company_type.value_counts().nlargest(9).index)), 'pref_company_type'] = 'Unsure/Others'
finaldf.pref_company_type.value_counts()
```

For the career_interest column, we standardise the 4 possible answers as Yes, No, Not answered or Unsure/others. We first replace the binary values with their string counter parts and the nan values with Not answered.

```python
finaldf.career_interest.replace(np.nan, 'Not answered', inplace = True)
finaldf.career_interest.replace(1, 'Yes', inplace = True)
finaldf.career_interest.replace(0, 'No', inplace = True)
```

Since we are only interested in whether the respondents are interested in a job related to computer science, we change all values to one of the 4 options mentioned above. The Unsure/others values encompasses all values that was typed in by respondents rather than given as an option.

```python
listofyes = ['Full-Stack Web Developer', 'Front-End Web Developer', 'Data Scientist', 'Game Developer', 'Back-End Web Developer', 'Information Security', 'Mobile Developer', 'Data Engineer', 'DevOps / SysAdmin', 'User Experience Designer', 'Product Manager', 'Teacher / Trainer / Developer Evangelist', 'Quality Assurance Engineer']
finaldf.loc[finaldf.career_interest.str.contains('I am not interested in a software development career'), 'career_interest'] = 'No'
finaldf.loc[~finaldf.career_interest.isin(list(finaldf.career_interest.value_counts().nlargest(16).index)), 'career_interest'] = 'Unsure/Other options'
finaldf.loc[finaldf.career_interest.isin(listofyes), 'career_interest'] = 'Yes'
```

```python
finaldf.career_interest.value_counts()
```

For the employment_field column, since we are only interested in whether the respondents worked in computer science related fields, we will categorise them by whether they are working, and if they are working in computer science related fields.

```python
finaldf.employment_field.replace(np.nan, 'Not answered', inplace = True)
finaldf.employment_field = finaldf.employment_field.str.capitalize()
finaldf.loc[finaldf.employment_field.str.contains('Software'), 'employment_field'] = 'Yes'
finaldf.loc[finaldf.employment_field == 'I am not currently working', 'employment_field'] = 'Not Employed'
finaldf.loc[~finaldf.employment_field.isin(['Not answered', 'Yes', 'Not employed']), 'employment_field'] = 'No'
finaldf.employment_field.value_counts()
```

Since the expected_annual_inc_from_sw_dev and annual_inc columns are of similar formatting, we shall clean them together. For 2016-2018, they asked for the exact numeric values but in 2021, they gave ranges for respondants to choose from. Since replacing the values in 2021 with the mean of the bin ranges changes the interquartile range by a lot (due to the number of data points), they shall simply be replaced with nan.

```python
finaldf.expected_annual_inc_from_sw_dev = pd.to_numeric(finaldf.expected_annual_inc_from_sw_dev, errors = 'coerce')
finaldf.annual_inc = pd.to_numeric(finaldf.annual_inc, errors = 'coerce')
```

Next, we will replace any outliers in both columns with their respective medians, since the number of such occurreces are low. The range of outliers will be values outside the range of 0 to 250000 since those are the lower and upper bounds for the bins given by the 2021 dataset.

```python
finaldf.loc[(finaldf.expected_annual_inc_from_sw_dev < 0) | (finaldf.expected_annual_inc_from_sw_dev > 250000), 'expected_annual_inc_from_sw_dev'] = finaldf.expected_annual_inc_from_sw_dev.median()
finaldf.loc[(finaldf.annual_inc < 0) | (finaldf.annual_inc > 250000), 'annual_inc'] = finaldf.annual_inc.median()
```

Based on the value_count of the age column, the most likely age range of respondents is 8 to 80 years old. The rest will be replaced with median since the number of such occurrences is low.

```python
finaldf.loc[(finaldf.age < 8) | (finaldf.age > 80), 'age'] = finaldf.age.median()
```

For the gender columns, since we are looking for biological gender, we will categorise all the other answers under one category.

```python
finaldf.gender.replace(np.nan, 'Not answered', inplace = True)
finaldf.loc[~finaldf.gender.isin(['Male', 'Female', 'Prefer not to say', 'Not answered']), 'gender'] = 'Not biological gender'
finaldf.gender.value_counts()
```

For the education_level column, there were several formatting issues, such as capitalisation and use of different types of apostrophes and the formatting of High School diplomas in previous years being different.

```python
finaldf.education_level.replace(np.nan, 'Not answered', inplace = True)
finaldf.education_level = finaldf.education_level.str.capitalize()
finaldf.education_level = finaldf.education_level.str.strip()
finaldf.loc[finaldf.education_level.str.contains('Bachelor'), 'education_level'] = 'Bachelor\'s degree'
finaldf.loc[finaldf.education_level.str.contains('Associate'), 'education_level'] = 'Associate\'s degree'
finaldf.loc[finaldf.education_level == 'Some high school', 'education_level'] = 'High school diploma or equivalent (ged)'
finaldf.education_level.value_counts()
```

Since we are only interested in whether the respondents majored in Computer Science related subjects, we will categorise them as such.

```python
finaldf.university_major.replace(np.nan, 'Not answered', inplace = True)
finaldf.loc[(finaldf.university_major.str.contains('computer')) | (finaldf.university_major.str.contains('Computer')) | (finaldf.university_major.str.contains('software')) | finaldf.university_major.str.contains('Software'), 'university_major'] = 'Computer Science related'
finaldf.loc[~(finaldf.university_major.isin(['Not answered', 'Computer Science related', 'I didn\'t attend a university'])), 'university_major'] = 'Other subjects'
finaldf.university_major.value_counts()
```

For the marital_status column, the capitalisation was changed over the years.
